```python
import pandas as pd
# practicas Ventas de papelería
#Pedro Martinez 0093
```

```python
productos = {
    "Producto": ["Carpeta", "Lapizero", "Saca puntas", "Escuadra", "Hojas de Maquina"],
    "Precio": [35, 10, 250, 8, 15],
    "Cantidad": [5, 10, 2, 15, 6]
}

df = pd.DataFrame(productos)

df["Total"] = df["Precio"] * df["Cantidad"]

print("=== VENTAS DE PAPELERÍA ===")
print("====Pedro Martinez NC 0093====")
print(df)
```

```python
print("Venta total:", df["Total"].sum())
print("====Pedro Martinez NC 0093====")
```
